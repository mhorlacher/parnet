```python
import torch
```

```python
A_logit = torch.tensor([7, 3, 2], dtype=torch.float32)
B_logit = torch.tensor([9, -2, 8], dtype=torch.float32)

A_prob = torch.nn.functional.softmax(A_logit, dim=0)
B_prob = torch.nn.functional.softmax(B_logit, dim=0)
```

```python
M_logit = torch.tensor(0.77, dtype=torch.float32)
M_prob = torch.sigmoid(M_logit)
M_prob
```

```python
C_prob = M_prob * A_prob + (1 - M_prob) * B_prob
C_prob
```

```python
A_logprob = A_logit - torch.logsumexp(A_logit, dim=0)
B_logprob = B_logit - torch.logsumexp(B_logit, dim=0)
```

```python
# using logprobs for A and B and logit for M
C_logprob = torch.log( M_prob * torch.exp(A_logprob) + (1 - M_prob) * torch.exp(B_logprob) )
torch.exp(C_logprob)
```

```python
max_logprob = torch.maximum(A_logprob, B_logprob)
logsumexp = max_logprob + torch.log(
    M_prob * torch.exp(A_logprob - max_logprob) + (1 - M_prob) * torch.exp(B_logprob - max_logprob)
)
torch.exp(logsumexp)
```

```python
def f(x, y):
    return torch.maximum(x, y) + torch.log(torch.exp(-torch.abs(x - y)))

torch.exp(f(A_logprob, B_logprob))
```
