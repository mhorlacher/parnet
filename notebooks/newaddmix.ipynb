```python
import torch

from parnet.layers import AdditiveMix
```

```python
class LambdaLayer(torch.nn.Module):
    def __init__(self, fun):
        super(LambdaLayer, self).__init__()
        self.fun  = fun
    def forward(self, *args, **kwargs):
        return self.fun(*args, **kwargs)
```

```python
layer = AdditiveMix(num_tasks=2, mix_coeff_layer=lambda _: torch.tensor([0.5, 0.5]).unsqueeze(-1))

# Prepare inputs. We provide log-probs as logits, as they are easily generated from probs.
layer.head_target = LambdaLayer(
    lambda _: torch.tensor([[0.1, 0.5, 0.4], [0.2, 0.3, 0.5]]).log().unsqueeze(0)
)  # -> [0.15, 0.35, 0.4]
layer.head_control = LambdaLayer(
    lambda _: torch.tensor([[0.2, 0.4, 0.4], [0.1, 0.8, 0.1]]).log().unsqueeze(0)
)  # -> [0.15, 0.55, 0.3]
layer.mix_coeff = LambdaLayer(lambda _: torch.tensor([0.5, 0.5]).unsqueeze(0))
```
